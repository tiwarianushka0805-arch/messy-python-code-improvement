```python

students = [
    ["Rahul", 78, 88, 92],
    ["Priya", 65, 71, 69],
    ["Amit", 90, 94, 85],
    ["Sneha", 55, 60, 58],
    ["Vikram", 82, 79, 88]
]


def calculate_total(student):
    """Return the total marks of a student."""
    return sum(student[1:])


def calculate_grade(average):
    """Return a grade based on the student's average."""
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"


print("REPORT")

total_averages = 0
topper_name = ""
highest_average = 0

for student in students:
    name = student[0]
    total_marks = calculate_total(student)
    average = total_marks / 3
    grade = calculate_grade(average)

    total_averages += average

    if average > highest_average:
        highest_average = average
        topper_name = name

    print(f"{name} {total_marks} {average} {grade}")

class_average = total_averages / len(students)

print(f"Class average: {class_average}")
print(f"Topper: {topper_name} {highest_average}")
```
